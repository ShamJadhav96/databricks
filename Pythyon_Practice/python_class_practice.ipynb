```python
class Account():
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    def deposit(self,deep_amt):
        self.balance += deep_amt
        print(f"Available amount is after deposit {self.balance}")
    def withdrow(self,wit_amt):
        if self.balance >= wit_amt:
            self.balance -= wit_amt
            print(f"Amount has been debitaed and current balance is {self.balance}")
        else:
            print("Inificient Funds")
    def __str__(self):
        print(f"owner : {self.owner}\nbalance: {self.balance}")
```

```python
a=Account("Sham",500)
```
