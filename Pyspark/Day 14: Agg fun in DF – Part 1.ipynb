```python
#creating dataframe to do aggregate funtions handson

from pyspark.sql import *
Data = [
    Row(id = 1, value = 10),
    Row(id = 2, value = 20),
    Row(id = 3, value = 30),
    Row(id = 4, value = 10),
    Row(id = 5, value = 50),
    Row(id = 6, value = None),
    Row(id = 7, value = 80)
]

df = spark.createDataFrame(Data)
display(df)
```

```python
#Summation (sum): Sums up the values in a specified column
from pyspark.sql import functions as f

sum_df = df.select(f.sum("value")).show()
```

```python
#average of the values in a specified column

from pyspark.sql import functions as f
avg_df = df.select(f.round(f.avg("value"),2)).show()
```

```python
#Count (count): Counts the number of non-null values in a specified column.
from pyspark.sql import functions as f
count_df = df.select(f.count("value")).show()
```

```python
#Maximum (max) and Minimum (min): Finds the maximum and minimum values in a specified column.

from pyspark.sql import functions as f
maxmin_df = df.select(f.max("value"),f.min("value")).show()
```

```python
#chceking is valuye column is null or not
from pyspark.sql import functions as f
isnull_df = df.select(f.isnull("value")).show()
```

```python
#Distinct Values Count (countDistinct): Counts the number of distinct values in a specified column.
from pyspark.sql import functions as f

distinct_df = df.select(f.countDistinct("value")).show()
```

- **Handling Nulls:**  
  The `count` function will count only non-null values, while `sum`, `avg`, `max`, and `min` will ignore null values in their calculations.

- **Performance:**  
  Aggregate functions can be resource-intensive, especially on large datasets. Using the appropriate partitioning can improve performance.

- **Use Cases:**  
  - **Summation:** Useful for calculating total sales, total revenue, etc.  
  - **Average:** Helpful for finding average metrics like average sales per day.  
  - **Count:** Useful for counting occurrences, such as the number of transactions.  
  - **Max/Min:** Helps to determine the highest and lowest values, such as maximum sales on a specific day.  
  - **Distinct Count:** Useful for finding unique items, like unique customers or products.
