### Loading Data from CSV File into a DataFrame
- Loading data into DataFrames is a fundamental step in any data processing workflow in PySpark. 
- This document outlines how to load data from CSV files into a DataFrame, including using a custom schema and the  implications of using the inferSchema option.

```python
#Import Required Libraries
from pyspark.sql import SparkSession #import *
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType #import *

#Defining schema
schema_define  = StructType([
    StructField("EmployeeID", IntegerType(), True),
    StructField("Name", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("Salary", IntegerType(), True),
    StructField("Age", IntegerType(), True),
    StructField("City", StringType(), True),
    StructField("DepartmentID", IntegerType(), True)
])

#reading CSV file
df_def_schema = spark.read.csv("/Volumes/workspace/default/practice_files/employees.csv", header=True, schema = schema_define)

#result
df_def_schema.display()
```

```python
#printing Schema
df_def_schema.printSchema()
```

### Load Multiple CSV Files
- To read multiple CSV files into a single DataFrame, you can pass a list of file paths.
- Ensure that the schema is consistent across all files.

```python
#Import Required Libraries
from pyspark.sql import SparkSession #import *
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType #import *

#Defining schema
schema_def  = StructType([
    StructField("EmployeeID", IntegerType(), True),
    StructField("Name", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("Salary", IntegerType(), True),
    StructField("Age", IntegerType(), True),
    StructField("City", StringType(), True),
    StructField("DepartmentID", IntegerType(), True)
])

multi_file_paths = [
    "/Volumes/workspace/default/practice_files/employees.csv",
    "/Volumes/workspace/default/practice_files/employees_2.csv",
    "/Volumes/workspace/default/practice_files/employees_3.csv"
    ]

#reading CSV file
df_Multi = spark.read.csv(multi_file_paths, header=True, schema = schema_def)

#result
df_Multi.display()

#We can display by this way also
#display(df_Multi)
```

### 📌 inferSchema in PySpark
`inferSchema=True` allows Spark to automatically detect the data type of each column while reading a CSV file.

**How it Works**
- Spark scans the CSV file to determine the appropriate data type for each column.
- This requires an additional scan before loading the data.

**Advantages**
- ✅ No need to define the schema manually.
- ✅ Useful when the schema is unknown or changes frequently.

**Disadvantages**
- ❌ Slower for large datasets due to the extra scan.
- ❌ May infer incorrect data types if the data is inconsistent.
- ❌ Less control compared to using an explicit schema.

**Best Practice**
- **Development / Learning:** Use `inferSchema=True`.
- **Production:** Define the schema explicitly for better performance and consistency.
