For 3+ years of experience, interviewers usually move beyond basic PySpark syntax and focus more on Spark internals, performance optimization, joins, data handling, and real-world troubleshooting.

🔥 Most common PySpark interview questions

1. PySpark & Spark basics  
   - What is PySpark and why do we use it?  
   - What is the difference between RDD, DataFrame, and Dataset?  
   - What is lazy evaluation?  
   - What are transformations and actions?  
   - What is the difference between narrow and wide transformations?  
   - What causes a shuffle in Spark?  
   - What is a Spark job, stage, and task?  
   - Explain Spark execution flow.  
   - What is the role of the driver and executors?  
   - What is Spark's DAG?  

2. DataFrame operations  
   - Difference between select(), selectExpr() and withColumn().  
   - Difference between filter() and where().  
   - How do you remove duplicates?  
   - How do you handle null values?  
   - How do you rename multiple columns?  
   - How do you cast a column to another datatype?  
   - How do you create a new column based on conditions?  
   - How do you find the latest record for each customer?  
   - How do you find duplicate records?  
   - How do you get the top N records per group?  

3. Joins  
   - Explain different types of joins in PySpark.  
   - What is a broadcast join?  
   - When would you use broadcast()?  
   - What happens if the broadcast table is too large?  
   - What is a shuffle join?  
   - How do you optimize a slow join?  
   - What is data skew?  
   - How do you identify data skew?  
   - How do you handle skewed joins?  
   - Difference between broadcast join and sort-merge join.  

4. Partitioning & performance  
   - What is partitioning in Spark?  
   - Difference between repartition() and coalesce().  
   - When would you use repartition()?  
   - When would you use coalesce()?  
   - How do you determine the optimal number of partitions?  
   - What is partition pruning?  
   - What is predicate pushdown?  
   - How can you optimize a PySpark job?  
   - What causes excessive shuffling?  
   - How do you troubleshoot a slow Spark job?  

5. Cache & persistence  
   - What is cache()?  
   - What is persist()?  
   - Difference between cache() and persist().  
   - When should you cache a DataFrame?  
   - When should you not cache?  
   - What is unpersist()?  

6. Window functions  
   - What is a window function?  
   - Difference between row_number(), rank() and dense_rank().  
   - Find the latest transaction for each customer.  
   - Find the second-highest salary in each department.  
   - Calculate running total using a window.  
   - Find previous/next record using lag() and lead().  

7. Reading & writing data  
   - How do you read CSV/JSON/Parquet files?  
   - Why is Parquet preferred over CSV?  
   - How do you define an explicit schema?  
   - What happens if schema inference is enabled?  
   - How do you write partitioned data?  
   - Difference between overwrite and append.  
   - How do you handle corrupt records?  
   - How do you optimize file sizes?  

8. Spark SQL  
   - How do you use SQL with PySpark DataFrames?  
   - Difference between groupBy() and window functions.  
   - How do you register a temporary view?  
   - Difference between createOrReplaceTempView() and a permanent table.  
   - Can you mix PySpark DataFrame API and Spark SQL?  

9. Real-world scenario questions  
   These are particularly important at 3+ years:  
   - Your PySpark job is taking 2 hours instead of 20 minutes. How will you troubleshoot it?  
   - One partition has significantly more data than others. What is the problem?  
   - Your join is causing a huge shuffle. How will you optimize it?  
   - You have a 500 GB fact table and a 50 MB dimension table. How would you join them?  
   - Your Spark job is running out of memory. What could be the reason?  
   - You have millions of small files. How would you handle them?  
   - How would you optimize a DataFrame with many transformations?  
   - How would you handle data skew in a real project?  
   - How would you process an incremental load using PySpark?  
   - How would you handle duplicate records in an ETL pipeline?
