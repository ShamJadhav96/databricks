Overview  
• **Purpose:** Both union and unionAll are used to combine two DataFrames into a single DataFrame.  
• **DataFrame Compatibility:** The two DataFrames must have the same schema (i.e., the same column names and data types) to perform the union operation.

```python
'''
# syntax
union_df = df1.union(df2)
unionall_df = df1.unionAll(df2)
'''
```

```python
from pyspark.sql import Row

# Create two DataFrames with the same schema
df1 = spark.createDataFrame([Row(id=1, name='Alice'), Row(id=2, name='Bob')])
df2 = spark.createDataFrame([Row(id=2, name='Bob'), Row(id=3, name='Charlie')])

# Perform union and unionAll
union_df = df1.union(df2)
unionall_df = df1.unionAll(df2)

display(union_df)
display(unionall_df)
```

**union()**
- Functionality:  
  Combines two DataFrames and retains all rows, including duplicate rows from the result.
- Behavior:  
  The union() method does not retain only unique rows across both DataFrames, resulting in a DataFrame with duplicates.

**unionAll()**
- Functionality:  
  Combines two DataFrames and retains all rows, including duplicates.
- Behavior:  
  The unionAll() method performs the union operation but does not eliminate duplicate rows, similar to union().

```python
unique_df = union_df.dropDuplicates()
display(unique_df)
```

```python
display(unionall_df)
```

```python
uniqueall_df = unionall_df.distinct()
display(uniqueall_df)
```

```python
# Notes for union and unionAll in PySpark:
# - Both union() and unionAll() combine two DataFrames with identical schemas.
# - union() and unionAll() retain all rows, including duplicates.
# - unionAll() is deprecated; use union() instead.
# - To remove duplicates after union, use dropDuplicates() or distinct().
```
