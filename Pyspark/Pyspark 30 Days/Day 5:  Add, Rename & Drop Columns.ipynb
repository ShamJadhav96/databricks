### Add, Rename & Drop Columns


**_You can rename columns using the alias() method:_**

```python
from pyspark.sql.functions import *

df = spark.read.csv("/Volumes/workspace/default/practice_files/employees.csv", header=True, inferSchema=True)
```

```python
df1= df.select(
        col("Name").alias('EmployeeName'), #renamig column name from Name to EmployeeName
        col("Salary").alias('EmployeeSalary'), #renaming column salary to EmployeeSalary
        column("Department"), #selecting column Department
        df.DepartmentID #Selecting column DepartmentID
    )
```

```python
df1.display()
```

**_Using selectExpr() for Concise Column Selection_**

```python
#selectExpr() allows you to use SQL expressions directly and rename columns concisely:
df2 = df.selectExpr("Name As EmployeeName", "Salary as EmployeeSalary", "Department")
df2.display()
```

• **_Use col(), column(), or string names to select columns._**

• **_Use expr() and selectExpr() for SQL-like expressions and renaming._**

• **_Use alias() to rename columns._**

• **_Get the list of columns using df.columns._**

### PySpark DataFrame Manipulation part 2: Adding, Renaming, and Dropping Columns

**_Adding New Columns with withColumn()_**


In PySpark, the withColumn() function is widely used to add new columns to a DataFrame.
You can either assign a constant value using lit() or perform transformations using existing
columns.


```python
#Add a constant value column:
newdf = df.withColumn("NewColumn", lit("Sham"))
newdf.display()
```

```python
newdf1 = df.withColumn("withCountry", expr("City = 'Pune'"))
newdf1.display()
```

This function allows adding multiple columns, including calculated ones:

• Example:

    o Assign a constant value with lit().
    o Perform calculations using existing columns like multiplying values.

**_Renaming Columns with withColumnRenamed()_**


PySpark provides the withColumnRenamed() method to rename columns. This is especially
useful when you want to change the names for clarity or to follow naming conventions:

```python
newdf2 = df.withColumnRenamed("Age","    Employee_Age &")
newdf2.show()
```

```python
#Handling column names with special characters or spaces: If a column has special
#characters or spaces, you need to use backticks (`) to escape it:
newdf2.select("    Employee_Age &").show()
```

**_Dropping Columns with drop()_**

```python
#To remove unwanted columns, you can use the drop() method:
df_drop = newdf.drop("NewColumn")
df_drop.show()
```

```python
df_drop1 = newdf1.drop("withCountry", "Age")
df_drop1.show()
```

_Dropping columns creates a new DataFrame, and the original DataFrame remains
unchanged_.

**_Immutability of DataFrames_**

In Spark, DataFrames are immutable by nature. This means that after creating a DataFrame,
its contents cannot be changed. All transformations like adding, renaming, or dropping
columns result in a new DataFrame, keeping the original one intact.

    • For instance, dropping columns creates a new DataFrame without altering the original:
    newdf = df.drop("ItemType", "SalesChannel")
This immutability ensures data consistency and supports Spark’s parallel processing, as
transformations do not affect the source data.

    • Use withColumn() for adding columns, with lit() for constant values and expressions for computed values.

    • Use withColumnRenamed() to rename columns and backticks for special characters or spaces.

    • Use drop() to remove one or more columns.

    • DataFrames are immutable in Spark—transformations result in new DataFrames, leaving the original unchanged.
