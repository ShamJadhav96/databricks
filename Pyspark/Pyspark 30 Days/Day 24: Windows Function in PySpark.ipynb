**Windows Function in PySpark**

- **Window functions** in PySpark:
  - Perform calculations across a set of rows related to the current row within a specified partition.
  - Unlike `groupBy` functions:
    - Do **not** reduce the number of rows in the result.
    - Calculate a value for **each row** based on the specified window.

### Importing required libraries

###practice

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.window import Window\
    
spark=SparkSession.builder.appName("Sham").getOrCreate()

data = [
    ("A", "X", 1, "2023-01-01"),
    ("A", "X", 2, "2023-01-02"),
    ("A", "Y", 3, "2023-01-01"),
    ("A", "Y", 3, "2023-01-02"),
    ("B", "X", 5, "2023-01-01"),
    ("B", "X", 4, "2023-01-02")
]

columns = ["category", "sub_category", "value", "timestamp"]
df = spark.createDataFrame(data,columns)

window_sepc = Window.partitionBy("category","sub_category").orderBy(col("timestamp"),col("value"))

df = df.withColumn("Row_Number", row_number().over(window_sepc))
df = df.withColumn("Rank", rank().over(window_sepc))
df = df.withColumn("Danse_rank", dense_rank().over(window_sepc))
df = df.withColumn("Lead_next_row", lead("value").over(window_sepc))
df = df.withColumn("Lag_previos_row", lag("value").over(window_sepc))
df = df.withColumn("Avg", avg("value").over(window_sepc))

df.show()
```


**Notes on Above Cells:**

- Window functions allow calculations across rows related to the current row, without reducing row count.
- Common window functions: `row_number()`, `rank()`, `dense_rank()`, `lead()`, `lag()`, and aggregate functions like `avg()`, `sum()`, `min()`, `max()`.
- Window specification defines partitioning and ordering for calculations.
- Example demonstrates usage of multiple window functions on a sample DataFrame.
- Useful for tasks like ranking, finding previous/next values, and computing aggregates within groups.
