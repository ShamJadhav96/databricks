**Union and UnionByName**

```python
from pyspark.sql import SparkSession

saprk = SparkSession.builder.appName("UnionandUnionByName").getOrCreate()

data = [
    (1,"Sham","Pune"),
    (2,"Om","Daund"),
    (3,"Pankaj","Mumbai"),
    (4,"Hyndavi","Hydrabad"),
    (5,"Pinky","Indore"),
    (6,"Pratik","Dharashiv")
]

schema = (["ID","Name","City"])
emp1 = spark.createDataFrame(data,schema)
emp1.show()

data1 = [
    (2,"Om",10000),
    (3,"Pankaj",9000),
    (7,"Sidram",7000),
    (4,"Hyndavi",100000),
    (5,"Pinky",100000),
    (8,"Ram",80000),
    (9,"Akash",70000)
]
schema1 = (["Id","Name","Salary"])
emp2 = spark.createDataFrame(data1,schema1)
emp2.show()
```

```python
'''union_df = emp1.union(emp2)
display(union_df)
'''

#above line of code will create issue so we need to use unionByName

unionByName_df = emp1.unionByName(emp2, allowMissingColumns = True)
display(unionByName_df)
```

```python
from pyspark.sql.functions import *

result_df = unionByName_df.groupBy("ID").agg(
    first("Name", ignorenulls=True).alias("Name"),
    first("City", ignorenulls=True).alias("City"),
    first("Salary", ignorenulls=True).alias("Salary")
)
result_df = result_df.dropDuplicates()
display(result_df)
```


### Quick Revision Notes: Union and UnionByName in PySpark

- **Union**: Combines two DataFrames with the same schema (column names and types must match).
- **UnionByName**: Combines DataFrames by matching column names, allows missing columns (`allowMissingColumns=True`).

#### Example:
- `emp1` and `emp2` have different columns.
- `emp1.union(emp2)` fails due to schema mismatch.
- `emp1.unionByName(emp2, allowMissingColumns=True)` works, aligning columns by name.

#### Aggregation:
- Use `groupBy` and `agg` to merge rows by ID and select first non-null values for each column.
- `first("col", ignorenulls=True)` picks the first non-null value in each group.

#### Useful Functions:
- `dropDuplicates()` removes duplicate rows.
- `display()` shows DataFrame output in Databricks.

---
**Summary:**  
Use `unionByName` for DataFrames with different columns. Aggregate to combine info by key (e.g., ID).
