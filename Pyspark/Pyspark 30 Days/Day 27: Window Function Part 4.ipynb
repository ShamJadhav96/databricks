**highest Salary in department**

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.window import Window

spark = SparkSession.builder.appName("WindowPractice").getOrCreate()
# Updated sample data for employees
emp_data = [
    (1, "Alice", 1, 6300),
    (2, "Bob", 1, 6200),
    (3, "Charlie", 2, 7000),
    (4, "David", 2, 7200),
    (5, "Eve", 1, 6300),
    (6, "Frank", 2, 7100)
]

# Sample data for departments
dept_data = [
    (1, "HR"),
    (2, "Finance")
]

emp_Schema = ["Id","Name","dept_id","Salary"]
dept_Schema =["dept_id","department"]

emp_df = spark.createDataFrame(emp_data,emp_Schema)
dept_df = spark.createDataFrame(dept_data,dept_Schema)

emp_df.show()
dept_df.show()
```

```python
window_spec = Window.partitionBy("dept_id").orderBy(desc("Salary"))

df = emp_df.withColumn("Rank",rank().over(window_spec))
result_df = df.alias("a").join(dept_df.alias("b"), df.dept_id == dept_df.dept_id, "inner")\
    .filter(df.Rank==1)\
    .select("a.Name", "a.dept_id", "a.Salary", "b.department")
display(result_df)
```
