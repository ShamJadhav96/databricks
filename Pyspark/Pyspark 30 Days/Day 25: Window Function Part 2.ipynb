**Windows Function in PySpark Part 2**

- Define a window specification  
- Using `rank()` to calculate rank  
- Using `dense_rank()` to calculate dense rank  
- Using `row_number()` to calculate row number  
- Using `lead()` to calculate the difference with the next row  
- Using `lag()` to calculate the difference with the previous row

```python
# creating sample data for window fun
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.window import Window

spark = SparkSession.builder.appName("Timepass").getOrCreate()

data = [
    ("Alice", 850),
    ("Bob", 920),
    ("Charlie", 780),
    ("David", 990),
    ("Eve", 670),
    ("Frank", 880),
    ("Grace", 940),
    ("Hannah", 760),
    ("Ivy", 830),
    ("Jack", 910),
    ("Kate", 940),
    ("Leo", 780),
    ("Mona", 850),
    ("Nina", 880),
    ("Oscar", 670)
]

columns = ["name","marks"]
df = spark.createDataFrame(data,columns)
window_spec = Window.orderBy("marks")
```

```python
df = df.withColumn("Rank",rank().over(window_spec))
df.show()
```

```python
#Using dense_rank() to calculate dense rank

df = df.withColumn("dense_rank", dense_rank().over(window_spec))
df.show()
```

```python
#Using row_number() to calculate row number

df = df.withColumn("Row_Number",row_number().over(window_spec))
df.show()
```

```python
#Using lead() to calculate the difference with the next row
df=df.withColumn("Score Diff with next person",lead("marks").over(window_spec)-df["marks"])
df.show()
```

```python
#Using lag() to calculate the difference with the previous row

df = df.withColumn("Diff in score with previous person", df["marks"]-lag("marks").over(window_spec))
df.show()
```
