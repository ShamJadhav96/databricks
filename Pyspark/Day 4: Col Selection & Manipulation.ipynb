### Different Methods to Select Columns

```python
# In PySpark, you can select specific columns in multiple ways:
#  Using col() function/ column() / string way:

from pyspark.sql.functions import *

df = spark.read.csv("/Volumes/workspace/default/practice_files/employees.csv", header=True, inferSchema=True)
```

```python
df.select(col("Name")).show()
```

```python
df.select(column("Salary")).show()
```

```python
df.select("Name").show()
```

```python
df.select(df.Age).show()
```

```python
#Selecting Multiple Columns Together
#You can combine different methods to select multiple columns:

df2 = df.select(col("Name"), column("Salary"), "Age", df.City)
```

```python
df2.display()
```
