### Defining Schema Programmatically with StructType

```python
from pyspark.sql.types import *


#Defining Schema using StructType and structField
empschema = StructType ([
    StructField("EmployeeID", IntegerType(), True),
    StructField("Name", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("Salary", IntegerType(), True),
    StructField("Age", IntegerType(), True),
    StructField("City", StringType(), True),
    StructField("DepartmentID", IntegerType(), True)
])

#creating dataframe using manual schema defining.
df_employee = spark.read.load("/Volumes/workspace/default/practice_files/employees.csv", format ="csv", header =True, schema=empschema)

df_employee.display()
```

```python
df_employee.printSchema()
```

### Define the schema as a string

```python
#Defining Schema in String 
emp_string_schema = '''
EmployeeID Integer,
Name String,
Department String,
Salary Integer,
Age Integer,
City String,
DepartmentID Integer
'''
#reading data using defined schema
df_employee_String = spark.read.load("/Volumes/workspace/default/practice_files/employees.csv", format="csv", header=True, schema=emp_string_schema)

df_employee_String.display()
```

```python
df_employee_String.printSchema()
```
