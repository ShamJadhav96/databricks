**_Changing Data Types (Schema Transformation)_**

In PySpark, you can change the data type of a column using the cast() method. This is
helpful when you need to convert data types for columns like Salary or Phone.



```python
from pyspark.sql.functions import *

df = spark.read.load("/Volumes/workspace/default/practice_files/employees.csv", format="csv", header=True, inferSchema=True)
df.show()
```

```python
#In this we are changing data type of salary column 
cast_df = df.withColumn("Salary", col("Salary").cast("double"))
cast_df.show()
```

```python
#here we are assigning the original data type to salary column.
cast_df = cast_df.withColumn("Salary", col("Salary").cast("Integer")) 
cast_df.show() 
```

**_we are doing for multiple columns_**

```python
#for multiple column we added one extra row here.
cast_df = cast_df.withColumn("phone",lit(9999999999))  
```

```python
#for multiple column
cast_df = (cast_df.withColumn("phone",col("phone").cast("Integer"))
                  .withColumn("Salary",col("Salary").cast("double"))
)
```

```python
#reverting changes
cast_df = (cast_df.withColumn("phone",col("phone").cast("long"))
                  .withColumn("Salary",col("Salary").cast("integer"))
)
```

**_Filtering on Null or Non-Null Values_**

```python
from pyspark.sql.types import *
#creating null dataset
emp_schema = StructType([
    StructField("EmployeeID", IntegerType(), True),
    StructField("Name", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("Salary", IntegerType(), True),
    StructField("Age", IntegerType(), True),
    StructField("City", StringType(), True),
    StructField("DepartmentID", IntegerType(), True)
])

emp_data = [
    (101, "Shyam",  "IT",        75000, 27, "Pune",      1),
    (102, "Rahul",  "HR",        55000, 29, "Mumbai",    2),
    (103, "Amit",   "Finance",   68000, 31, "Delhi",     3),
    (104, "Priya",  "IT",        82000, 26, "Bangalore", 1),
    (105, "Neha",   "Marketing", None,  28, "Hyderabad", 4),   # Salary NULL
    (106, "Rohit",  None,        58000, 30, "Chennai",   5),   # Department NULL
    (107, None,     "Finance",   72000, 32, "Kolkata",   3),   # Name NULL
    (108, "Vikas",  "IT",        90000, None, "Pune",    1),   # Age NULL
    (109, "Meena",  "HR",        50000, 25, None,        2),   # City NULL
    (110, "Sneha",  "Marketing", 65000, 29, "Mumbai",    None) # DepartmentID NULL
]

emp_df = spark.createDataFrame(emp_data, emp_schema)

emp_df.show()
```

```python
#checking for null values
df_null = emp_df.filter(col("Salary").isNull())
df_null.show()
```

```python
#checking for notnullvalues

df_notnull = emp_df.filter(col("Age").isNotNull())
df_notnull.show()
```

**_Counting Distinct Values_**

```python
# here we need to result using print statement as count is returning the int value 
# Count distinct values in the 'Department' column

df_distinct_count = df.select(col("DepartmentID")).distinct().count()
print(df_distinct_count)

# also we can do one more thing so we cn use show ()

df_distinct_show = df.select(countDistinct("DepartmentID").alias("Distinct_dep_count")).show()
```

```python
# taking distinct count by using 3 combine column 
print("taking distinct count by using 3 combine column")
df_distinct_multiple = df.select(col("Name"), "Age", df["DepartmentID"]).distinct().count()
print(df_distinct_multiple)

#taking distinct count using show method
print("taking distinct count using show method")
df_dis_mul_show = df.select(countDistinct(col("Name"), "Age", df["DepartmentID"]).alias("mulcount"))
df_dis_mul_show.show()

#taking multiple distinct count for seperate columns
print("taking multiple distinct count for seperate columns")
df_multi = df.select(countDistinct("Name").alias("Name_count"),
                     countDistinct("Age").alias("Age_count"),
                     countDistinct("DepartmentID").alias("DepartmentID_count")
)
df_multi.show()
```

5. _Unique/Distinct Values: Use .distinct() to get unique rows or distinct values in a
