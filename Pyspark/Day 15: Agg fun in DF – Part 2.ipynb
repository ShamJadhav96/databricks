```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("AggregationPractice").getOrCreate()

schema  = StructType([
    StructField("department", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("bonus", IntegerType(), True),
    StructField("employee_name", StringType(), True)
])

Data =[
    ("HR", 10000, 500, "John"),
    ("Finance", 20000, 1500, "Doe"),
    ("HR", 15000, 1000, "Alice"),
    ("Finance", 25000, 2000, "Eve"),
    ("HR", 20000, 1500, "Mark")
]

df = spark.createDataFrame(Data,schema)
df.show()
```

```python
#Grouped Aggregation: Perform aggregation within groups based on a grouping column.

from pyspark.sql import functions as f
group_agg_df = df.groupBy("department").agg(
    f.sum("salary"),
    f.avg("salary"),
    f.max("salary"),
    f.min("salary")
)
display(group_agg_df)
```

```python
# Multiple Aggrigation

from pyspark.sql import functions as f

multi_agg_df = df.groupBy("department").agg(
    f.count("salary"),
    f.avg("bonus"),
    f.max("salary")
).show()
```

**Concatenate Strings**

    Concatenate strings within a column.

concat_ws: Concatenates string values within the column, separating them by the
specified delimiter (, ).

```python
from pyspark.sql import functions as f
agg_df = df.agg(f.concat_ws(", ",f.collect_list("employee_name")).alias("Conctaed_Names")).show(truncate=False)
```

```python
#here we are using group by

from pyspark.sql import functions as f
agg_grop_df = df.groupby("department").agg(f.concat_ws(", ",f.collect_list("employee_name")).alias("Conctaed_Names")).show(truncate=False)
```

**First and Last**
    Find the first and last values in a column (within each group).

```python
from pyspark.sql import functions as f
first_last_df = df.groupBy("department").agg(
    f.first("employee_name"), 
    f.last("employee_name")
    ).show()
```

**Standard Deviation and Variance**
Calculate the standard deviation and variance of values in a column.

    • stddev: Calculates the standard deviation of column.
    • variance: Calculates the variance of column.


```python
from pyspark.sql import functions as f

std_ver_df = df.select(f.stddev("salary"),f.variance("salary")).show()
```

```python
from pyspark.sql import functions as f
group_agg_alias_df = df.groupBy("department").agg(
    f.sum("salary").alias("Sum_Salary"),
    f.avg("salary").alias("Avg_Salary"),
    f.max("salary").alias("Max_Salary"),
    f.min("salary").alias("Min_Salary")
)
display(group_agg_alias_df)
```
