**_Sample Data Creation_**

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("Sortingadstring").getOrCreate()

sales_schema = StructType([
    StructField("Country", StringType(), True),
    StructField("Region", StringType(), True),
    StructField("UnitsSold", IntegerType(), True),
    StructField("UnitPrice", DoubleType(), True)
])

Data = [
    ("USA", "North America", 100, 50.5),
    ("India", "Asia", 300, 20.0),
    ("Germany", "Europe", 200, 30.5),
    ("Australia", "Oceania", 150, 60.0),
    ("Japan", "Asia", 120, 45.0),
    ("Brazil", "South America", 180, 25.0)
]

df_sales = spark.createDataFrame(Data,sales_schema)
df_sales.show()
```

**_Sorting the DataFrame_**


```python
#Sort by a single column (ascending order):
df = df_sales.orderBy("Country").show()
```

```python

#Sort by a multiple column (Ascending Order):
df1 = df_sales.orderBy("Country","UnitsSold").show()

#Here, the DataFrame is sorted first by Country (ascending), and within the same country, it is sorted by UnitsSold in ascending order
```

**_Sort by a column in descending order and limit:_**

```python
#Sort by a single column (descending Order)
df2 = df_sales.orderBy(desc("Country")).show()

#Sort by a multiple column (descending Order)
df3 = df_sales.orderBy(desc("Country"), desc("UnitsSold")).show()
```

**_Sorting with null values last:_**


```python
df4 = df_sales.orderBy(desc("Country"), null_last=True).show()
#This ensures that null values (if present) are placed at the end when sorting by Country
```

### String Functions

**_initcap_**

```python
#Convert the first letter of each word to uppercase (initcap):
#df = df_sales.select(initcap(col("Country"))).show()

df = df_sales.select(initcap("Country")).show()
```

**_Lower_**

```python
#Convert all text to lowercase (lower):

df = df_sales.select(lower("Country")).show()
df = df_sales.select(lower(col("Country")).alias("lower")).show() # just aligning the name to column
```

**_upper_**

```python
#Convert all text to lowercase (Upper):
df = df_sales.select(upper("Country")).show()
df = df_sales.select(upper(col("Country")).alias("UPPER")).show() # just aligning the name to column
```

### Concatenation Functions

**_Concatenate two columns_**

```python
#cancating two columns country and region

df = df_sales.select(concat(col("Country"),col("Region")).alias("Country & Region")).show()

# Concatenates the values of Region and Country without any separator.
```

**_Concatenate with a separator:_**

```python
#Concatenates the values of Region and Country with | as a separator.

df = df_sales.select(concat_ws(' |',col("Country"),col("Region")).alias("Country & Region")).show()
```

```python
Concatanated_df = df_sales.withColumn("Country & Region", concat(col("Country"),lit(" "), concat(col("Region")))).show()
```


**_String Manipulation: You can convert strings to lowercase, uppercase, or capitalize
the first letter of each word. Use initcap(), lower(), and upper() for these
transformations._**

**_Concatenation: Use concat() to join two columns or concat_ws() to join with a
separator._**
