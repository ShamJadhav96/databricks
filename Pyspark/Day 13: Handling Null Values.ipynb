```python
data = [
    ("John", "North", 100, None),
    ("Doe", "East", None, 50),
    (None, "West", 150, 30),
    ("Jane", None, 200, 40),
    ("Smith", "North", None, None),
    (None, None, None, None),
    (None, None, 200, 40),
    ("Sham", "South", 150, 36)
]

columns = ["first_name", "region", "sales", "discount"]
df = spark.createDataFrame(data, columns)
df.display()
```

**Detecting Null Values:**
The isNull() function identifies rows where a specified column has null values. The
output shows a boolean flag for each row to indicate whether the value in the
column is null.

```python
from pyspark.sql.functions import *
df.select(col("first_name"),col("region"),isnull(col("first_name")).alias("First_Name_null"),isnull(col("region")).alias("Null_Regions")).display()
```

**Dropping Rows with Null Values:**

- `dropna()` removes rows that contain null values in any column when the default mode is used.
- Specifying `"all"` ensures rows are only removed if all columns contain null values.
- You can also apply null handling only on specific columns by providing a list of column names to the `subset` parameter.

```python
df1 = df.dropna()
display(df1)
```

```python
df2 = df.dropna("all")
display(df2)
```

```python
df3 = df.dropna(how="all", subset=["first_name", "region"])
display(df3)
```

**Filling Null Values:**
- `fillna()` allows replacing null values with specified replacements, either for all columns or selectively.
- In the example, nulls in `region` are replaced with `"Unknown"`, while `sales` and `discount` nulls are filled with `0`.

```python
df4 = df.fillna({"region":"Unknown","sales":0, "discount":0})
display(df4)
```

```python
df5 = df.fillna("N/A", subset = ["first_name", "region"])
display(df5)
```

**Coalesce Function:**

The `coalesce()` function returns the first non-null value in a list of columns.  
It’s useful when you need to handle missing data by providing alternative values from other columns.

```python
df6 = df.withColumn("Sales_Profit_aggrigate", coalesce("sales","discount"))
display(df6)
```

**Handling Nulls in Aggregations:**

- Null values can distort aggregate functions like `mean()`.
- Using `coalesce()` in an aggregation ensures that any null values are replaced with a default (e.g., `0.0`) to avoid skewing the results.

```python
df7 = df.groupBy("region").agg(coalesce(mean("sales"), lit(0))).alias("Aggregated_Sales")
display(df7)
```

**Null Handling in DataFrames - Summary**

1. **Detecting Nulls:**  
   Use `isNull()` to identify null values in specific columns.

2. **Dropping Nulls:**  
   `dropna()` removes rows with null values, either in any or all columns.  
   You can target specific columns using the `subset` parameter.

3. **Filling Nulls:**  
   `fillna()` replaces nulls with specified default values, either for all or selected columns.

4. **Coalesce Function:**  
   `coalesce()` returns the first non-null value from multiple columns, providing a fallback when some columns contain nulls.

5. **Aggregations:**  
   Use `coalesce()` during aggregations like `mean()` to handle nulls by substituting them with defaults (e.g., `0`), ensuring accurate results.
