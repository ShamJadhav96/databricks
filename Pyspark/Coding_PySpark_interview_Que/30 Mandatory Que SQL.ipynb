```python
%sql
--deduplication
WITH cte AS (
    SELECT *,
           ROW_NUMBER() OVER(PARTITION BY emp_id ORDER BY emp_id) AS rn
    FROM Employees
)
DELETE FROM cte
WHERE rn > 1;
```

```python
%sql
--Running total
SELECT
    emp_id,
    emp_name,
    salary,
    SUM(salary) OVER (ORDER BY emp_id ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_total
FROM employee;
```
