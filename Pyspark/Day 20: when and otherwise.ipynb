**when**:  
The `when` function in PySpark is used to define a condition. If the condition is met, it returns the specified value. You can chain multiple `when` conditions to handle various cases.

**otherwise**:  
The `otherwise` function specifies a default value to return if none of the conditions in the `when` statements are met.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

spark = SparkSession.builder.appName("New_Way").getOrCreate()

emp_data = [
    (1, "Sham", 4000, 28),
    (2, "Rahul", 3500, 25),
    (3, "Priya", 2000, 30),
    (4, "Amit", 5000, 27),
    (5, "Sneha", 4500, 32),
    (6, "Vikas", 1000, 22),
    (7, "Neha", 6000, 29),
    (8, "Rohit", 3500, 24),
    (9, None, 4000, 31),
    (10, "Kiran", None, 26)
]
columns = StructType([
    StructField("ID",IntegerType(),True),
    StructField("Sham",StringType(),True),
    StructField("Salary",IntegerType(),True),
    StructField("Age",IntegerType(),True)
])

emp = spark.createDataFrame(emp_data, columns)
emp.show()
```

```python
emp = emp.withColumn("Status",when(col("Age")<25, "Young").when((col("Age")>25) & (col("age")<35), "Adult").otherwise("Aged"))\
    .withColumn("Buget", when(col("Salary")<=2000,"low").when((col("Salary")>2000)&(col("Salary")<=4000),"medium").otherwise("Heigh"))
emp.show()
```

Explanation

1. **"status" column**:  
   - Assigns "Young" if age < 30  
   - Otherwise assigns "Adult"

2. **"income_bracket" column**:  
   - Assigns "Low" if salary < 4000  
   - Assigns "Medium" if salary is between 4000 and 4500  
   - Assigns "High" for any other salary values

This approach allows for flexible handling of multiple conditions in PySpark DataFrames using `when` and `otherwise`.
