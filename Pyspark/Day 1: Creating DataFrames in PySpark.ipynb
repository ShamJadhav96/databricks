### PySpark: Spark SQL and DataFrames
- Spark SQL is a module in Spark for working with structured data. It allows you to query structured data inside Spark using SQL and integrates seamlessly with DataFrames.
- A DataFrame in PySpark is a distributed collection of data organized into named columns.
- It’s conceptually similar to a table in a relational database or a DataFrame in R/Python.

_**Creating Dataframe using straightforward wayusing hardcoded values.**_


```python
#Sample Data 
data =  [(1,"John"), (2,"Mary"), (3,"Jane"), (4,"Bob"), (5,"Alice"), (6,"Steve"), (7,"Julia"), (8,"Paul"), (9,"Sue"), (10,"Kim")]

#column names
columns = ["ID", "Name"]

#create dataframe using data and columnn
df = spark.createDataFrame(data, columns)

#Show Dataframe
#df.show()
df.display()
```

_**Creating DataFarme using Pandas**_

```python
import pandas as pd 

#creating sample panda dataframe
pandas_df = pd.DataFrame(data, columns=columns)

#Creating dataframe
df_pandas_dataframe = spark.createDataFrame(pandas_df)
#df_pandas_dataframe.show()

#display(df_pandas_dataframe)
df_pandas_dataframe.display()
```

**_Creating DataFrame using Dictionary_**

```python
#passing data dictionary
data_dic= [{"ID":1, "Name": "Sham"},{"ID":2, "Name": "Ravi"}]

#creating dataframe
df_dic = spark.createDataFrame(data_dic)

#display(df_dic)
df_dic.show()
```

**_Creating empty DataFrame_**

```python
#import StructType StructField
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

#define schema here
schema = StructType([
    StructField("ID",IntegerType(),True),
    StructField("Name",StringType(),True)
])

#Creating Dataframe
df_empty = spark.createDataFrame([],schema)
df_empty.show()
```

_**To Pass datain empty frame**_

```python
#import StructType StructField
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

#define schema here
schema = StructType([
    StructField("ID",IntegerType(),True),
    StructField("Name",StringType(),True)
])

data = [(1,"Sham"),
        (2,"Om")
    ]

#Creating Dataframe
df_empty = spark.createDataFrame(data,schema)
df_empty.show()
```

**The show() function in PySpark displays the contents of a DataFrame in a tabular format. It
has several useful parameters for customization:**

_1. n: Number of rows to display (default is 20)_

_2. truncate: If set to True, it truncates column values longer than 20 characters (default
is True)._

_3. vertical: If set to True, prints rows in a vertical format._

```python
df.show(n=3, truncate=25, vertical=True)
#Show entire DataFrame (default settings):
df.show()
#Show the first 10 rows:
df.show(10)
#Show DataFrame without truncating any columns:
df.show(truncate=False)
```
