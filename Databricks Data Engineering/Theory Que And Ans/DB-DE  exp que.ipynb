**Very likely scenario:**  
*"Your Databricks pipeline has multiple stages involving slowly changing dimensions (SCD), deduplication, selecting the latest record per customer, and top N scenarios."*

- **Slowly Changing Dimensions (SCD):**  
  - Example: Find the latest transaction for every customer.
  - Source contains both new and updated customers.
  - Use Delta MERGE to update existing customers and insert new customers into the Delta target table.

- **Troubleshooting Questions (Real-world):**
  - Scenario 1: Pipeline succeeded but the target table contains 0 records. What will you check?
  - Scenario 2: Pipeline is running very slowly. What will you check?
  - Scenario 3: One customer has 10 million records and other customers have only 100 records. Your join is slow. Why?  
    - *Data skew.*
  - Scenario 4: Your Delta table has millions of small files. What will you do?  
    - *OPTIMIZE / compaction and review ingestion/write strategy.*
  - Scenario 5: Your pipeline failed halfway. How will you restart without creating duplicates?  
    - *Idempotent processing + checkpoints + MERGE/business keys + appropriate workflow retry design.*
  - Scenario 6: A source column changed from integer to string. What happens?  
    - *Discuss schema enforcement/evolution depending on the ingestion mechanism and configuration.*
  - Scenario 7: User accidentally deleted records from a Delta table. Can you recover them?  
    - *Delta time travel / restore, subject to retention and vacuum considerations.*
