
1. Databricks architecture & project questions  
Absolutely. For a 3.6-year Databricks Data Engineer interview, I would answer these in a way that sounds like you have actually worked on pipelines—not like textbook definitions.  
You can use the following answers almost directly in the interview.

1. Explain the architecture of your Databricks project.  
**Answer:**  
"In my project, we follow a Medallion architecture with Bronze, Silver and Gold layers. Data comes from source systems and is ingested into ADLS using Auto Loader/Lakeflow. We store the raw data in Bronze Delta tables. In Silver, we perform cleansing, deduplication, data validation and business transformations. Gold contains the final business-ready tables used for reporting and analytics. We use Unity Catalog for governance and access control, and Databricks Workflows for orchestration and scheduling."

2. What components of Databricks have you worked on?  
**Answer:**  
"I have mainly worked with Delta Lake, Unity Catalog, Databricks Workflows, Lakeflow, Auto Loader, PySpark and SQL. I have used Delta tables for storing data, Auto Loader for incremental file ingestion, Lakeflow for building data pipelines, Workflows for orchestration, and Unity Catalog for data governance and access control. I have also worked on Spark performance optimization and job monitoring."

3. Explain your end-to-end data pipeline in Databricks.  
**Answer:**  
"The pipeline starts when new files arrive in ADLS. We use Auto Loader or Lakeflow to incrementally ingest those files into Bronze Delta tables. From Bronze, we process the data into Silver where we handle cleansing, nulls, duplicates, data types and business transformations. Then we aggregate and transform the Silver data into Gold tables. The entire process is orchestrated through Databricks Workflows, and Unity Catalog manages the tables and permissions."

**Simple flow:**  
Source  
   ↓  
ADLS  
   ↓  
Auto Loader / Lakeflow  
   ↓  
Bronze Delta  
   ↓  
Cleaning + Deduplication + Validation  
   ↓  
Silver Delta  
   ↓  
Business Transformations + Aggregations  
   ↓  
Gold Delta  
   ↓  
Reporting / Analytics  

4. How did data come into Databricks?  
**Answer:**  
"The source data was mainly available as files in ADLS. For incremental file ingestion, we used Auto Loader. It identifies and processes only new files instead of scanning and processing all the existing files repeatedly. We then wrote the incoming data into Bronze Delta tables."

If they ask "Why Auto Loader?", say:  
"Because it is scalable for continuously arriving files and provides checkpointing and schema evolution capabilities."

5. Where did you store raw, transformed and curated data?  
**Answer:**  
"We followed the Medallion architecture. Raw or minimally processed data was stored in Bronze Delta tables, cleansed and transformed data in Silver Delta tables, and business-ready or aggregated data in Gold Delta tables. These tables were registered and governed through Unity Catalog."

6. Why did you use Delta Lake instead of Parquet?  
**Answer:**  
"The main reason is that Delta provides features that plain Parquet doesn't provide, such as ACID transactions, schema enforcement, schema evolution, time travel and MERGE support. This is especially useful for incremental processing and updating existing records. Delta also provides better reliability for production data pipelines."

If they ask:  
"Give me an example."  
Say:  
"For example, if customer records arrive every day and some existing customers are updated, instead of rewriting the complete table, we can use Delta MERGE to update existing records and insert new records."

7. Explain your Bronze → Silver → Gold architecture.  
**Answer:**  
"Bronze is mainly the raw ingestion layer, so we try to preserve the source data with minimal transformation. Silver is our cleaned and standardized layer where we handle data quality, duplicates, nulls, data types and business rules. Gold contains business-level tables, aggregations and KPIs that are directly consumed by reporting or downstream applications."

A good interview line:  
"Bronze is source-oriented, Silver is transformation-oriented, and Gold is business-oriented."

8. What transformations did you perform in Silver?  
**Answer:**  
"In Silver, we performed data cleansing and standardization. For example, we handled null values, converted data types, removed duplicates, standardized date formats, applied business validations, filtered invalid records and performed joins with reference data. We also implemented incremental processing and MERGE logic wherever required."

If they ask for an example:  
"For example, if the source had duplicate customer records, I would use a window function based on customer ID and update timestamp to keep the latest record before writing it to the Silver table."

9. How did you handle incremental data?  
**Answer:**  
"For file-based ingestion, we used Auto Loader with checkpointing so that already processed files weren't processed again. For table-level incremental processing, we used keys and timestamps or Delta Change Data Feed where applicable. For updates and inserts, we used Delta MERGE based on the business key."

**Example:**  
MERGE INTO target t  
USING source s  
ON t.customer_id = s.customer_id  
WHEN MATCHED THEN  
  UPDATE SET *  
WHEN NOT MATCHED THEN  
  INSERT *  

Then say:  
"This allows us to process only the new or changed records instead of reprocessing the complete dataset."

10. How did you handle failures and rerun your pipeline?  
**Answer:**  
"We designed the pipeline to be restartable and idempotent. We used checkpoints for streaming or Auto Loader processing, and MERGE logic for incremental updates. In Workflows, we configured retries for transient failures. If a task failed, we checked the error from the job run and Spark logs, fixed the issue and reran the failed task or workflow rather than unnecessarily processing everything again."

If they ask:  
"What is idempotency?"  
"It means running the same pipeline again should not create duplicate or incorrect data."

11. How did you monitor your Databricks pipelines?  
**Answer:**  
"We used Databricks Workflows to monitor job and task status. We checked run duration, task failures, logs and Spark UI for performance issues. For production failures, we also configured notifications or alerts so that the team could take action. For data quality, we monitored record counts, rejected records and validation failures."

12. How did you optimize your Spark jobs?  
**Answer:**  
"First, I check the Spark UI to identify where the job is spending time, especially shuffle-heavy stages and skewed partitions. Then I optimize joins using broadcast joins where appropriate, reduce unnecessary shuffles, select only required columns, filter data as early as possible and optimize partitioning. For Delta tables, I also look at small-file problems and use OPTIMIZE where appropriate."

If they ask "What performance problems have you faced?", a good answer is:  
"One common issue is data skew, where a few keys contain a very large amount of data. This can make one or two partitions much slower than the others. I identify this through Spark UI and then consider techniques like repartitioning, salting or broadcasting a smaller dataset depending on the use case."
