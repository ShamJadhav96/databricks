Data Visualization

```python
#a) Display Category wise profit as a Bar Chart

import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/vijay6909/Sales-Project-/refs/heads/master/Sample%20-%20Superstore.csv"

df = pd.read_csv(url)

category_profit = df.groupby('Category')['Profit'].sum()

plt.figure(figsize=(6,4))
plt.bar(category_profit.index, category_profit.values, color='skyblue')
plt.title('Category Wise Profit')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.show()
```

```python
#b) Segment wise Profit as a Pie Chart
import pandas as pd
import matplotlib.pyplot as plt

segment_profit = df.groupby('Segment')['Profit'].sum()

plt.figure(figsize=(6,6))
plt.pie(
    segment_profit,
    labels=segment_profit.index,
    autopct='%1.1f%%'
)
plt.title('Segment Wise Profit')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df['Year'] = pd.to_datetime(df['Order Date'], format='mixed').dt.year
year_profit = df.groupby('Year')['Profit'].sum()

plt.figure(figsize=(8,4))
plt.plot(
    year_profit.index,
    year_profit.values,
    marker='o'
)
plt.title('Profit by Year')
plt.xlabel('Year')
plt.ylabel('Profit')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(
    df['Sales'],
    df['Profit'],
    alpha=0.5
)
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

url = "https://raw.githubusercontent.com/rushabh-mehta/EDA-on-Automobile-Dataset/refs/heads/master/Automobile_data.csv"
df = pd.read_csv(url)


df.replace('?', pd.NA, inplace=True)
for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors='ignore')
df.dropna(subset=['price'], inplace=True)
df_numeric = df.select_dtypes(include='number')
df_numeric = df_numeric.fillna(df_numeric.mean())

X = df_numeric.drop('price', axis=1)
y = df_numeric['price']

# Split data: 80% Train, 20% Test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

#predict price for 20% test data

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Determine R-Squared value
r2 = r2_score(y_test, y_pred)

print("R-Squared Value:", round(r2, 4))

# Compare Actual vs Predicted
result = pd.DataFrame({
    "Actual Price": y_test,
    "Predicted Price": y_pred
})

print("\nSample Predictions:")
print(result.head(10))
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score

# Load dataset
url = "https://raw.githubusercontent.com/sudipta002/Pima_Indians_Binary_Classification_Keras/refs/heads/main/pima_indians_diabetes.csv"
df = pd.read_csv(url)

X = df.iloc[:, :-1]
y = df.iloc[:, -1]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Predict on test data
y_pred = model.predict(X_test)

# Classification Report
print(classification_report(y_test, y_pred))

# F1 Score
f1 = f1_score(y_test, y_pred)

print("F1 Score:", round(f1, 4))
```
