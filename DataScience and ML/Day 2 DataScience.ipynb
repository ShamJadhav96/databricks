```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ordersdf = pd.read_csv('https://raw.githubusercontent.com/vijay6909/Sales-Project-/refs/heads/master/Sample%20-%20Superstore.csv')
 
ordersdf
```

```python
ordersdf.shape
ordersdf.info()
ordersdf.describe(include="all")
```

```python
columns_without_spaces = [x.replace(' ', '') for x in ordersdf.columns]
columns_without_spaces
ordersdf.columns = columns_without_spaces
ordersdf
```

```python
# convert date columns from string to datetime
ordersdf['OrderDate'] = ordersdf['OrderDate'].astype('datetime64[ns]')
ordersdf['ShipDate'] = ordersdf['ShipDate'].astype('datetime64[ns]')
ordersdf.dtypes
```

```python
# create a pivot table with segment in rows and category in columns and profilt, sales in values
orders_pivot = pd.pivot_table(
    ordersdf,
    index='Category',
    columns='Segment',
    values=['Profit', 'Sales'],
    aggfunc=np.sum)
orders_pivot
 
```

```python
# Display Category wise revenue as a bar plot
CategoryRevenueDF = ordersdf.groupby('Category')['Sales'].sum().sort_values(ascending=False)
 
plt.figure(figsize=(4, 3))
plt.bar(CategoryRevenueDF.index, CategoryRevenueDF.values)
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.title("Category wise Revenue")
plt.show()
```

```python
# find the top 10 sub categories by sales
top10SubCatSales = ordersdf.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)
print(top10SubCatSales)
plt.figure(figsize=(10,5))
plt.barh(top10SubCatSales.index[::-1], top10SubCatSales.values[::-1])
plt.xlabel("Sales")
plt.ylabel("Sub-Category")
plt.title("Top 10 Sub-Category by Sales")
plt.show()
```

```python
# Generate a Pie plot showing Segment wise counts
plt.figure(figsize=(4, 3))
plt.pie(ordersdf['Segment'].value_counts(), labels=ordersdf['Segment'].value_counts().index, autopct='%1.1f%%')
plt.title("Segment wise Counts")
```

```python
# Display Cumulative sales over time using an Area chart
dailySalesDF = ordersdf.groupby('OrderDate')['Sales'].sum().reset_index()
dailySalesDF['CumulativeSales'] = dailySalesDF['Sales'].cumsum().round()
 
plt.figure(figsize=(8, 4))
plt.fill_between(dailySalesDF['OrderDate'], dailySalesDF['CumulativeSales'], alpha=0.5)
plt.xlabel("Date")
plt.ylabel("Cumulative Sales")
plt.title("Cumulative Sales over time")
# plt.xticks(rotation=45)
plt.xticks(fontsize=9)
plt.show()
```

```python
# Using seaborn library
# Generate Region wise sales as a Box plot
 
sns.boxplot(x='Region', y='Sales', data=ordersdf)
plt.title("Region wise Sales")
plt.show()
```

```python
# Monthly sales trend
ordersdf["salesMonth"] = ordersdf["OrderDate"].dt.strftime("%Y-%m")
monthlySalesdf = ordersdf.groupby('salesMonth')['Sales'].sum().reset_index()
 
plt.figure(figsize=(10, 5))
sns.lineplot(x='salesMonth', y='Sales', data=monthlySalesdf)
plt.title("Monthly Sales")
plt.xticks(rotation=90)
plt.xticks(fontsize=9)
plt.show()
```

```python
# Generate a corelation heat map comparing Sales, Profit, Discount, Quentity
 
corelationDF = ordersdf[['Sales', 'Profit', 'Discount', 'Quantity']].corr()
# corelationDF
 
sns.heatmap(corelationDF, annot=True, cmap='coolwarm')
plt.title("Feature Corelation")
plt.show()
 
```

```python
# Display sales by Region + Category
sns.catplot(x='Region', y='Sales', hue='Category', data=ordersdf, kind='bar', errorbar=None)
plt.show()  
```

```python
sns.jointplot(x='Profit', y='Sales', data=ordersdf, kind='reg')
plt.show()
```

```python
# Generate a Retail dashboard with 4 charts
plt.figure(figsize=(20, 10))
plt.subplot(2, 2, 1)
plt.scatter(ordersdf['Discount'], ordersdf['Profit'], alpha=0.5)
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.title("Discount vs Profit")
 
plt.subplot(2, 2, 2)
plt.bar(ordersdf['Segment'], ordersdf['Sales'], alpha=0.5)
plt.xlabel("Segment")
plt.ylabel("Sales")
plt.title("Sales by Segment")
 
plt.subplot(2, 2, 3)
sns.violinplot(x='Category', y='Discount', data=ordersdf)
plt.title("Discount by Category")
 
plt.subplot(2, 2, 4)
sns.boxplot(x='Region', y='Sales', data=ordersdf)
plt.title("Region wise Sales")
plt.show()
```
