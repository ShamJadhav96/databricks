```python
from scipy.stats import norm
 
mean = df['Sales'].mean()
std = df['Sales'].std()
 
lower = mean - std
upper = mean + std
 
print(mean, std)
 
df[(df['Sales']>=lower) & (df['Sales']<=upper)][['Order ID', 'Sales']]
```

```python
# According to Normal Distribution:
 
# 68% Rule : Approximately 68% of observations in a normally distributed dataset fall within ±1 Standard Deviation of the Mean.
 
# Cumulative Probability (CDF) => Cumulative Distribution Function of a Normal Distribution.
# It returns the probability that a random variable is less than or equal to x.
 
prob_1sd = (norm.cdf(mean+std, mean, std) - norm.cdf(mean-std, mean, std))
print(prob_1sd)
 
# 95% Rule : 95% within 2 SD
prob_2sd = (norm.cdf(mean+2*std, mean, std) - norm.cdf(mean-2*std, mean, std))
print(prob_2sd)
 
# 99.7% Rule : 99.7% within 3 SD
# Nearly all observations lie within ±3 Standard Deviations.
 
prob_3sd = (norm.cdf(mean+3*std, mean, std) - norm.cdf(mean-3*std, mean, std))
print(prob_3sd)
 
# Therefore the code is simply identifying:
# "Which retail orders fall into the expected normal sales range around the average?"
 
# Any orders beyond mean ± 2*std and mean ± 3*std are unusual
```

```python
mean = df['Sales'].mean()
std = df['Sales'].std()
 
lower_1sd = mean - std
upper_1sd = mean + std
 
print(lower_1sd, upper_1sd)
 
lower_2sd = mean - (2 * std)
upper_2sd = mean + (2 * std)
 
print(lower_2sd, upper_2sd)
 
unusual_orders = df[
    (df['Sales'] < lower_2sd) |
    (df['Sales'] > upper_2sd)
]
 
print(f"Total Orders : {len(df)}  Unusual Orders : {len(unusual_orders)}")
print(f"Percent of Unusual Orders : {round(len(unusual_orders) / len(df) * 100,2)}%")
print(unusual_orders[['Order ID','Sales']])
 
print(len(unusual_orders) / len(df) * 100)
# print(unusual_orders[['Order ID','Sales']])
```

```python
lower_3sd = mean - (3 * std)
upper_3sd = mean + (3 * std)
 
extreme_orders = df[
    (df['Sales'] < lower_3sd) |
    (df['Sales'] > upper_3sd)
]
 
print(f"Total Orders : {len(df)}  Extreme Orders : {len(extreme_orders)}")
print(f"Percent of Extreme Orders : {round(len(extreme_orders) / len(df) * 100,2)}%")
print(extreme_orders[['Order ID','Sales']])
```

```python
sample_df = df.sample(n=1000,  random_state=42)
sample_df1 = df.sample(frac=0.2,  random_state=42)
```

```python
# ### Confidence Interval
# #### What is the likely range of Average Sales?
# ##### Suppose you have sales data for 100,000 orders.
# ##### You don't want to analyze all orders, so you take a sample of 1000 orders and calculate:
# ##### Sample Mean Sales = ₹1,220
# ##### The sample mean is only an estimate of the true population mean.
#
# ##### Instead of saying => The average sales is ₹1,220
# ##### we say => We are 95% confident that the true average sales lies between ₹1,074 and ₹1,366
# ##### This range is called the 95% Confidence Interval.
```

```python
# For a population mean:
#  CI = sample Mean ± (Critical Value × Standard Error)
 
# Critical Value => Depends on confidence level.
# Confidence Level  Z Value
#    90%            1.645
#    95%            1.96
#    99%            2.576
 
import numpy as np
from scipy.stats import norm
 
mean = sample_df1['Sales'].mean()
std = sample_df1['Sales'].std()
n = len(sample_df1)
 
error_value = 1.96 * (std/np.sqrt(n))
 
lower = mean - error_value
upper = mean + error_value
 
print(lower,upper)
# 95% confidence that true average retail sales lies in this range.
 
```

```python
# **OUTLIER DETECTION**
# ###### Business wants to identify unusually large orders.
# ###### Method 1 : IQR (Interquartile Range) Method for outlier detection
#
# ###### Unlike the Z-score method, it does not assume the data is normally distributed, making it very useful for retail sales, profits, salaries, and other skewed business data.
 
# %%
q1 = df['Sales'].quantile(0.25)
q3 = df['Sales'].quantile(0.75)
iqr = q3 - q1
 
print("Q1 :" + str(q1) + "  Q3 :" + str(q3))
print("IQR (Q3 - Q1) : " + str(iqr))
 
# Outlier limits:
# Lower Limit = Q1 - 1.5 × IQR
# Upper Limit = Q3 + 1.5 × IQR
 
lower_limit = q1 - 1.5 * iqr
print("Lower Limit : " + str(lower_limit))
 
upper_limit = q3 + 1.5 * iqr
print("Upper Limit : " + str(upper_limit))
 
outliers = df[(df['Sales'] > upper_limit)]
print("Outlier Count : " + str(len(outliers)))
     
outliers[['Order ID', 'Sales']]
 
# Interpretation
# These are exceptional orders requiring further investigation.
```

```python
# ##### Method 2: Outlier detection using Z-Score
# ###### Formula:
# ###### Z=(X-μ)/σ
# ###### Where:
# ###### X = Value
# ###### μ = Mean
# ###### σ = Standard Deviation
# Rule : |Z| > 3 ⇒ Outlier
```

```python
from scipy.stats import zscore
 
df['ZScore'] = zscore(df['Sales'])
 
outliers = df.loc[abs(df['ZScore']) > 3, ['Order ID', 'Sales', 'Discount', 'Profit']]
print(len(df.loc[abs(df['ZScore']) > 3]))
print(outliers)
```
